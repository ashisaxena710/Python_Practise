```python
# Count the number of uppercase, lowercase, digits, and special characters in a string.
# Input:
# "Data@123"

# Output:
# Uppercase : 1
# Lowercase : 3
# Digits : 3
# Special Characters : 1

str_1 = input()
is_upper = 0
is_lower = 0
is_digit = 0
is_special = 0
for char in str_1:
    if char.isupper():
        is_upper += 1
    elif char.islower():
        is_lower += 1
    elif char.isdigit():
        is_digit += 1
    else:
        is_special += 1
print("Uppercase :",is_upper)
print("Lowercase :",is_lower)
print("Digits :",is_digit)
print("Special Characters :",is_special)
```

```python
# Find the longest word in a sentence.
# Input:->"Python is a powerful programming language"
# Output:->programming
str_1 = input()
str_2 = str_1.split(" ")
dict_1 = {}
for i in str_2:
    dict_1[i]= len(i)
max_length = max(dict_1.values())
for k,v in dict_1.items():
    if v== max_length:
        print(k)
```

```python
# Rotate a list to the right by K positions.
# Input:
#     list_1 = [1, 2, 3, 4, 5]
#     k = 2

# Output:->[4, 5, 1, 2, 3]
list_1 = [1, 2, 3, 4, 5]
k = 2
k = k % len(list_1)     
result = list_1[-k:] + list_1[:-k]
print(result)
```

```python
# Given a dictionary, print all keys whose values are greater than 50.
marks = {"Alice": 45,"Bob": 72,"Charlie": 90,"David": 38}
for k,v in marks.items():
    if v>50:
        print(k)
```

```python
# Find the missing number from a list containing numbers from 1 to n.
# Input:->[1, 2, 3, 5, 6]
# Output:->4

list_1 = eval(input())
for i in range(list_1[0],list_1[-1]+1):
    if i not in list_1:
        print(i)  
```
