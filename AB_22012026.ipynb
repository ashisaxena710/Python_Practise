```python
# Write a python program to check a given number is perfect or not.

num = int(input())

l=[]
for i in range(1,num):
    if num % i==0:
        l.append(i)
print(l)
print(sum(l))
if sum(l)==num:
#     print()
    print("It is Perfect Number")
else:
    print("It is not Perfect Number")
```

```python
# Palindrome 

string = input("")
if string==string[::-1]:
    print("String is a Palindrome")
else:
    print("String is not a Palindrome")
```

```python
# Write a Python program to swap two number without using third variable.

a=input()
b=input()

print("value of a is :",a)
print("value of b is :",b)

temp = b
b = a
a = temp

print("value of a is :",a)
print("value of b is :",b)
```

```python
# Write a Python program to calculate Euclidean Distance.
x_1, y_1 = 6, 7

x_2, y_2 = 5, 4

import math 
# euclidean distance = sqrt((x2-x1)^2 + (y2-y1)^2)
res = math.sqrt((x_2-x_1)**2 + (y_2-y_1)**2)
print("The Euclidean Distance between the above given two points  =",res)
```

```python
# Write a Python Program to Find Next Prime Number.

n=int(input())

for i in range(n,100000):
  for j in range(2,i):
    if i%j==0 :
      break
  else:
    print(f"The next prime number of {n} is: {i}")
    break
```
