```python
# Find the third largest number in a list without using sort() or max().
list_1 = [10, 50, 30, 20, 40]
first = float('-inf')
second = float('-inf')
third = float('-inf')

for num in list_1:
    if num> first:
        third = second
        second = first
        first = num
        
    elif num> second:
        third = second
        second = num
        
    elif num > third:
        third = num
        
print("Third largest : ",third)
```

```python
# Find the first non-repeating character in a string.
# input -> "swiss"
# output -> w
str_1 = input()

dict_1 = {}
for i in str_1:
    if i in dict_1.keys():
        dict_1[i] += 1
    else:
        dict_1[i] = 1
        
for k,v in dict_1.items():
    if v==1:
        print(k)
        break
```

```python
# Merge two dictionaries. If a key exists in both dictionaries, add their values.
# dict_1 = {"a":10, "b":20, "c":30}
# dict_2 = {"b":5, "c":15, "d":40}
# Output:-> {'a':10, 'b':25, 'c':45, 'd':40}

dict_1 = eval(input())
dict_2 = eval(input())
for k,v in dict_2.items():
    if k in dict_1.keys():
        dict_1[k] = dict_1[k] + v
    else:
        dict_1[k] = v
print(dict_1)
```

```python
# Check whether a list is sorted in ascending order.
# Input:->[2, 5, 8, 10]
# Output: ->Sorted

list_1 = eval(input())
is_sorted = True
for i in range(len(list_1)-1):
    if list_1[i] > list_1[i+1]:
        is_sorted = False
if is_sorted:
    print("Sorted")
else:
    print("Not Sorted")
```

```python
# Given a list of integers, separate the numbers into even and odd lists.
# Input:->[2, 7, 4, 9, 10, 13]

# Output:
#         Even : [2, 4, 10]
#         Odd  : [7, 9, 13]

list_1 = eval(input())
even = []
odd = []
for i in list_1:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print("Even :",even)
print("Odd :",odd)
```
