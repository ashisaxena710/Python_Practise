```python
# 1. Find the smallest number in a list without using min().
# Input:-> [12, 45, 3, 19, 8]
# Output:->3

list_1 = eval(input())
min_num = float('inf')
for num in list_1:
    if num < min_num:
        min_num = num
        
print(min_num)
```

```python
# 2. Check whether two strings are anagrams.
# Two strings are anagrams if they contain the same characters with the same frequency.
# Input:
#     listen
#     silent
# Output->Anagrams

str_1 = input("Str 1:")
str_2 = input("Str 2:")

print("---Method 1---")
if sorted(str_1)==sorted(str_2):
    print("Anagrams")
else:
    print("Not Anagrams")
    
print("\n---Method 2---")
from collections import Counter

if Counter(str_1)==Counter(str_2):
    print("Anagrams")
else:
    print("Not Anagrams")
print(Counter(str_1))
print(Counter(str_2))
```

```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if len(str1) != len(str2):
    print("Not Anagrams")
else:
    freq = {}

    for ch in str1:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    for ch in str2:
        if ch in freq:
            freq[ch] -= 1
        else:
            freq[ch] = -1

    if all(value == 0 for value in freq.values()):
        print("Anagrams")
    else:
        print("Not Anagrams")
```

```python
# Reverse the words in a sentence.
# Input:->"Data Analyst Interview"
# Output:->"Interview Analyst Data"
str_1 = input()
str_2 = str_1.split()
str_3 = str_2[::-1]
str_4 = " ".join(str_3)
print(str_4)
```

```python
# Given a dictionary of employee salaries, 
# print the names of employees whose salary is greater than the average salary.

# employees = {"Alice": 50000,"Bob": 70000,"Charlie": 65000,"David": 80000}

# output:
#     Bob
#     David

employees = {"Alice": 50000,"Bob": 70000,"Charlie": 65000,"David": 80000}
total_salary = sum(employees.values())
number_of_salary = len(employees)
avg_salary = total_salary/number_of_salary
# print(avg_salary)
for k,v in employees.items():
    if v>avg_salary:
        print(k)
```

```python
# Find all duplicate elements in a list.
# Input:->[1, 2, 3, 2, 4, 5, 3, 6]
# Output:->[2, 3]

list_1 = eval(input())
dict_1 = {}
for i in list_1:
    if i in dict_1.keys():
        dict_1[i] += 1
    else:
        dict_1[i] = 1
# print(dict_1)

duplicate_items = []
for k,v in dict_1.items():
    if v>1:
        duplicate_items.append(k)
print(duplicate_items)
```
