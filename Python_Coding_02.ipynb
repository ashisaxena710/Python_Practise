```python
# Write a Python program to check whether a number is prime.
num = int(input())
for i in range(2,num):
    if num%i==0:
        print("Not Prime")
        break
else:
    print("Prime")
```

```python
# Find the intersection of two lists (common elements only).
# list_a = eval(input("List 1: "))
# list_b =  eval(input("List 2: "))

list_a = [1,2,3,4,5]
list_b =  [4,5,6,7]

# method 1
common_elements = list(set(list_a).intersection(set(list_b)))
print(common_elements)

# method 2
common_Elements_ = [i for i in list_a if i in list_b]
print(common_Elements_)
```

```python
# Given a dictionary, find the key with the highest value without using max().
# Input-> marks = {"Alice": 85,"Bob": 92,"Charlie": 88}
# Output-> Bob : 92
dict_1 = eval(input())

max_v = float('-inf')
for v in dict_1.values():
    if v> max_v:
        max_v = v
print(max_v)

for k,v in dict_1.items():
    if v == max_v:
        print(f"{k}:{v}")
```

```python
# Count how many times each character appears in a string (ignore spaces).
# Input:->"data analyst"

# Output:
# d : 1
# a : 3
# t : 2
# ...
string_1 = input()
str_2 = string_1.split()
str_3 = "".join(str_2)

dict_1 = {}
for i in str_3:
    if i in dict_1.keys():
        dict_1[i] += 1
    else:
        dict_1[i] = 1

for k,v in dict_1.items():
    print(f"{k} : {v}")
```
