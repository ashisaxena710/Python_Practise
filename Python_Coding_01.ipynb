```python
# Write a Python program to find the largest number in a list without using the max() function.

# Input:
# [10, 45, 23, 89, 67]

# Output:
# 89
```

```python
numerical_list = [10,45,23,89,67]
numerical_list.sort()
print("1. largest num is ",numerical_list[-1])

numerical_list = [10,45,23,89,67]
largest_num = numerical_list[0]
for num in numerical_list:
    if num> largest_num:
        largest_num = num
print("2. largest num is ",largest_num)
```

```python
# Count the number of vowels and consonants in a string.
# Input:
# "Data Science"

# Output:
# Vowels: 5
# Consonants: 6
```

```python
str_1 = "Data Science"
str2 = str_1.split(" ")
str3 = "".join(str2)

vowels = []
consonants = []

for char in str3:
    if char in ["a","e","i","o","u"]:
        vowels.append(char)
    else:
        consonants.append(char)
        
print("Vowels:", len(vowels))
print("Cosonants:", len(consonants))
```

```python
# Remove duplicate elements from a list while maintaining the original order.

# input - [7, 3, 7, 1, 3, 9, 1]
# output - [7, 3, 1, 9]
```

```python
list_1 = eval(input())
result_list = []
for i in list_1:
    if i not in result_list:
        result_list.append(i)
print(result_list)        
```

```python
# Find the frequency of each word in a sentence.
# input - "python sql python excel sql python"

str1 = input()
str2 = str1.split()

dict_1 = {}

for i in str2:
    if i in dict_1.keys():
        dict_1[i] += 1
    else:
        dict_1[i] = 1
        
# print(dict_1)
for word, count in dict_1.items():
    print(f"{word} : {count}")
```

```python
employees = [
    {"name": "Alice", "salary": 50000},
    {"name": "Bob", "salary": 70000},
    {"name": "Charlie", "salary": 65000},
    {"name": "David", "salary": 80000}
]
# Without using max(), write a program to find:

# The employee with the highest salary.
# The average salary of all employees.
```

```python
employees = [
    {"name": "Alice", "salary": 50000},
    {"name": "Bob", "salary": 70000},
    {"name": "Charlie", "salary": 65000},
    {"name": "David", "salary": 80000}
]

# Initialize with the first employee
highest_employee = employees[0]
total_salary = 0

# Traverse the list
for employee in employees:
    total_salary += employee["salary"]

    if employee["salary"] > highest_employee["salary"]:
        highest_employee = employee

# Calculate average salary
average_salary = total_salary / len(employees)

# Print results
print("Employee with Highest Salary:")
print(highest_employee["name"], "-", highest_employee["salary"])

print("\nAverage Salary:")
print(average_salary)
```
